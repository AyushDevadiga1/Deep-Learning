```python
# Import necessary libraries

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt


from sklearn.datasets import make_classification
```

```python
# Experiment with make_classification function to create a linear seperable dataset 
X, y = make_classification(
                            n_samples=100, # No of rows
                            n_features=2, # No of columns
                            n_informative=1,
                            n_redundant=0,
                            n_classes=2, # No of classes in the dataset
                            n_clusters_per_class=1, 
                            random_state=41,
                            hypercube=False,
                            class_sep=14 # The Seperation distance between classes
                            )

# Experimented this with a lot of values and for most of them the model is unable to learn the best line distinguishing and in some cases even gives completely wrong decision 
# boundary , maybe this is why we use multiple layers of perceptrons that also with needed activation functions using the appropriate LOSS function to get the appropriate
# results and the best trained perceptron models
```

```python
# Plot the figure to visualize the dataset , do this recurively with above code block to get better results and then move on to check perceptron model .
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='Dark2',s=100)
plt.show()
```

```python
def perceptron(X,y):
    
    w1=w2=b=1
    lr = 0.1
    
    for j in range(1000):
    
        for i in range(X.shape[0]):

            # check condition
            z = w1*X[i][0] + w2*X[i][1] + b

            if z*y[i] < 0:
                w1 = w1 + lr*y[i]*X[i][0]
                w2 = w2 + lr*y[i]*X[i][1]
                b = b + lr*y[i]
            
    return w1,w2,b
            
```

```python
w1,w2,b = perceptron(X,y)
```

```python
m = -(w1/w2)
c = -(b/w2)

print(m,c)
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + c

plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='Dark2',s=100)
plt.ylim(-3,2)
plt.show()
```
