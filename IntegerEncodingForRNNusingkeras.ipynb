```python
import numpy as np
from keras.layers import TextVectorization
```

```python
data = [
            'wanna catch em all',
            'pokemon!!! GOOOO',
            'I am gonna catch em alllll',
            'Ash Ketchum born in Palet Town',
            'Ash encountered Pikachu',
            'Pikachu Iron TAIL!!!!!'
]
```

```python
# Instantiate the vectorizer layer
vectorizer = TextVectorization()

# Fit our custom data
vectorizer.adapt(data)
```

```python
# Now we can get the vocublary using get_vocabulary()
vectorizer.get_vocabulary() 
# The unknown string is "!"
```
