```python
# Import the necessary libraries
import numpy as np
import pandas as pd
```

```python
# Create the custom dataframe
df = pd.DataFrame(
                    [
                        [8,8,1],
                        [7,9,1],
                        [6,10,0],
                        [5,5,0]
                    ], columns=['cgpa', 'profile_score', 'placed'])
```

```python
# Initialize the parameters using this function ; same as we saw in regression

def initialize_parameters(layer_dims):

  np.random.seed(3)
  parameters = {}
  L = len(layer_dims)

  for l in range(1, L):

    parameters['W' + str(l)] = np.ones((layer_dims[l-1], layer_dims[l]))*0.1
    parameters['b' + str(l)] = np.zeros((layer_dims[l], 1))


  return parameters
```

```python
# Now we have another function here which is sigmoid unlike linear function only
# So we basically squeeze the real numbers into a range : 0 to 1

def sigmoid(Z):

  A = 1/(1+np.exp(-Z))

  return A
```

```python
# Same function but we pass on z to sigmoid for squashing and then return it
def linear_forward(A_prev, W, b):

  Z = np.dot(W.T, A_prev) + b

  # So basically we call the sigmoid function here to make sure that the output is the sigmoid here which is the probability instead of real number
  A = sigmoid(Z)

  return A
```

```python
# Forward Propogation using parameters that are made from the utiltiy function before and the matrix X as input

def L_layer_forward(X, parameters):

  A = X # Input to Input layer 1
  L = len(parameters) // 2 # number of layers in the neural network

  for l in range(1, L+1): # An additional loop to calculate the final output

    A_prev = A # In the current loop the previous loop's value becomes the previous_output

    # Fetching the parameters from the paramters hashmap using the layer index
    Wl = parameters['W' + str(l)]
    bl = parameters['b' + str(l)]

    # Pass the matrices to the linear forward utility function so it returns the output(Z)
    A = linear_forward(A_prev, Wl, bl)


  return A,A_prev # In the end the output will be returned with the output of the previous layer 
```

```python
# The classification parameter gradient had many terms due to the loss fuction : -y*np.log(y_hat) - (1-y)*np.log(1-y_hat) 
# This is binary cross entropy

def update_parameters(parameters, y, y_hat, A1, X, lr=0.001):
    # 1. Calculate the error term (negative sign ensures Gradient Descent)
    # This remains an unflattened object matching your input types
    error_term = lr * 2 * (y - y_hat)

    # 2. CACHE LOGIC: Store original W2 values before they are modified
    # .copy() prevents Python from auto-updating the cache when W2 changes
    w2_old = parameters['W2'].copy()
    w2_0_old = w2_old[0][0]
    w2_1_old = w2_old[1][0]

    # 3. Update Output Layer (W2, b2) 
    parameters['W2'][0][0] = parameters['W2'][0][0] - (error_term * A1[0][0])
    parameters['W2'][1][0] = parameters['W2'][1][0] - (error_term * A1[1][0])
    parameters['b2'][0][0] = parameters['b2'][0][0] - error_term # Fixed original b2 typo

    # 4. Update Hidden Layer Neuron 1 (W1 row 0, b1 row 0) using cached W2
    parameters['W1'][0][0] = parameters['W1'][0][0] - (error_term * w2_0_old * X[0][0])
    parameters['W1'][0][1] = parameters['W1'][0][1] - (error_term * w2_0_old * X[1][0])
    parameters['b1'][0][0] = parameters['b1'][0][0] - (error_term * w2_0_old)

    # 5. Update Hidden Layer Neuron 2 (W1 row 1, b1 row 1) using cached W2
    parameters['W1'][1][0] = parameters['W1'][1][0] - (error_term * w2_1_old * X[0][0])
    parameters['W1'][1][1] = parameters['W1'][1][1] - (error_term * w2_1_old * X[1][0])
    parameters['b1'][1][0] = parameters['b1'][1][0] - (error_term * w2_1_old)
```

```python
# Now lets propogate and backpropogate student by student 
```

```python
# Row 1 

X = df[['cgpa', 'profile_score']].values[0].reshape(2,1) # Shape(no of features, no. of training example)
y = df[['placed']].values[0][0]

# Parameter initialization
parameters = initialize_parameters([2,2,1])

print(f'\n Paramaters before : \n {parameters} \n')

y_hat,A1 = L_layer_forward(X,parameters)
y_hat = y_hat[0][0]

update_parameters(parameters,y,y_hat,A1,X)

print(f'\n Loss for this student = ',-y*np.log(y_hat) - (1-y)*np.log(1-y_hat))

print(f'\n Paramaters After : \n {parameters} \n')
```

```python
# Now looping it multiple times so that the model can learn parameters epoch by epoch

parameters = initialize_parameters([2,2,1])
epochs = 50

for i in range(epochs):

  Loss = []

  for j in range(df.shape[0]):

    X = df[['cgpa', 'profile_score']].values[j].reshape(2,1) # Shape(no of features, no. of training example)
    y = df[['placed']].values[j][0]

    # Parameter initialization


    y_hat,A1 = L_layer_forward(X,parameters)
    y_hat = y_hat[0][0]

    update_parameters(parameters,y,y_hat,A1,X)

    Loss.append(-y*np.log(y_hat) - (1-y)*np.log(1-y_hat))

  print('Epoch - ',i+1,'Loss - ',np.array(Loss).mean())
```
