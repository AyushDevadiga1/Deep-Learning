```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset,DataLoader
```

```python
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(
                            n_classes=2,
                            n_features=2,
                            n_informative=2,
                            n_redundant=0,
                            n_samples=100,
                            random_state=42
)
```

```python
# Now conver these to tensors
X = torch.tensor(X,dtype=torch.float32)
y = torch.tensor(y,dtype=torch.long)
```

```python
class CustomDataset(Dataset):
  def __init__(self,features,labels):
    self.features = features
    self.labels = labels
  def __len__(self):
    return self.features.shape[0]
  def __getitem__(self,indexes):
    return self.features[indexes],self.labels[indexes]
```

```python
dataset = CustomDataset(X,y)
```

```python
len(dataset) # our __len__ method does this
```

```python
dataset[34] # Our __getitem__ does this
```

```python
dataloader = DataLoader(dataset,batch_size=10,shuffle=True)
```

```python
for batch_features,batch_labels in dataloader: # Test whether the batches can be extracted or not.
  print(batch_features)
  print(batch_labels)
  print("-"*100)
```
