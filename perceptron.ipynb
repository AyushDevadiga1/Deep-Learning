```python
# This basically a practice session notebook on the perceptron concepts learned and to visualize why it does not work 
# on non linear datasets
```

```python
# Import Necessities

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import plotly.express as px
from mlxtend.plotting import plot_decision_regions

from sklearn.datasets import make_classification
from sklearn.datasets import make_circles

from sklearn.linear_model import Perceptron

```

```python
# Dataset 1 — Linear 

X_l, y_l = make_classification(
    n_samples=200,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_clusters_per_class=1,
    random_state=42
)
```

```python
# Dataset 2 — Non-linear

X_nl, y_nl = make_circles(
    n_samples=200,
    noise=0.1,
    factor=0.5,
    random_state=42
)
```

```python
# Now let's visualize them both

df_lin  = pd.DataFrame(X_l,  columns=["Feature 1", "Feature 2"])
df_lin["Class"] = y_l.astype(str)

df_circ = pd.DataFrame(X_nl, columns=["Feature 1", "Feature 2"])
df_circ["Class"] = y_nl.astype(str)

# ── Plot ─────────────────────────────────────────────────────────────
fig_lin = px.scatter(df_lin, x="Feature 1", y="Feature 2", color="Class",
                     title="Linear data — a straight line can separate these")

fig_circ = px.scatter(df_circ, x="Feature 1", y="Feature 2", color="Class",
                      title="Non-linear data — no straight line can separate these")

fig_lin.show()
fig_circ.show()
```

```python
# Train model for both the datasets
```

```python
p_l = Perceptron()
p_l.fit(X_l,y_l)

p_nl = Perceptron()
p_nl.fit(X_nl,y_nl)
```

```python
# Now lets plot both the model boundaries using ml_extend
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

plt.sca(axes[0])
plot_decision_regions(X_l, y_l, clf=p_l, legend=2)
axes[0].set_title(f"Linear data — Perceptron")
axes[0].set_xlabel("Feature 1")
axes[0].set_ylabel("Feature 2")

plt.sca(axes[1])
plot_decision_regions(X_nl, y_nl, clf=p_nl, legend=2)
axes[1].set_title(f"Non-linear data (circles) — Perceptron ")
axes[1].set_xlabel("Feature 1")
axes[1].set_ylabel("Feature 2")

plt.suptitle("Perceptron: Linear vs Non-linear data", fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Hence we concluded that perceptron works the best on linearly seperable data
```
