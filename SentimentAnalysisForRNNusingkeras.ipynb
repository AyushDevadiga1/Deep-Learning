```python
import numpy as np
import keras
from keras.layers import TextVectorization,Flatten,Dense
```

```python
data = [
            'wanna catch em all',
            'pokemon!!! GOOOO',
            'Gonna catch em alllll',
            'OH no !!! Pikachu !!!!',
            'This pokemon is crazy strong !!',
            'Ash Ketchum born in Palet Town'
]
```

```python
# Also add labels here so we will this dataset as a sample to use for sentiment analysis
labels = np.array([1,1,1,0,0,1])
```

```python
# Instantiate the vectorizer layer
vectorizer = TextVectorization(output_sequence_length=10) # The output_sequence_length itself is the padding here 

# Fit our custom data
vectorizer.adapt(data)
```

```python
# Now we can get the vocublary using get_vocabulary()
vectorizer.get_vocabulary() 
# The unknown string is "!"
```

```python
# To peek on what was the index assigned for each we do :
word_to_id = {word:idx for idx,word in enumerate(vectorizer.get_vocabulary())}

print("---Word to Integer Mapping---")
for word,idx in word_to_id.items():
    print(f'{word} : {idx}')
```

```python
# Get the padded sequnces from fitting vectorizing our data now
sequences = vectorizer(data)
```

```python
# Now we build the model for Integer Encodding

model = keras.Sequential(
    [
        keras.Input(shape=(10,)), # The input shape is same as what we passed above in padding
        Flatten(), # Flattent the layers
        Dense(10,activation='relu'), # Set the Hidden Layers
        Dense(1,activation='sigmoid')  # Set the Output Layer
    ]
)
```

```python
# Compile the model
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
# Train the model :
model.fit(sequences,labels,epochs=10,verbose=1)
```

```python
# As we see there is no accuracy change at all but that's not the point here 
```

```python
# Evaluate the model
loss,accuracy = model.evaluate(sequences,labels,verbose=0)
print(f'The loss for the model : {loss:.2f}')
print(f'The accuracy for the model : {accuracy:.2f}')
```
