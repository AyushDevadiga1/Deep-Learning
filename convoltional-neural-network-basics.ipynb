```python
# If n*n is the input and m*m is the kernal then feature_output = (n-m+1)*(n-m+1) # <--- WITHOUT any padding and striding = 1
```

```python
import tensorflow as tf
import keras 
import keras_tuner as kt

from keras.models import Sequential
from keras.layers import Input,Activation,Conv2D
from keras.layers import Dense,Flatten
from keras.datasets import mnist
```

```python
(X_train , X_test) , (y_train , y_test) = mnist.load_data()
```

```python
X_train.shape #<-- 2D we neet to flatten each row in 1d 
```

```python
# Model training without anything just normal striding = 1 , and simple convol
```

```python
model = Sequential()

# Add Input 
model.add(Input(shape=(28,28,1)))

n_layers = 3

# Define the 3 convol layers
for _ in range(n_layers):
    model.add(Conv2D(32,kernel_size=(3,3))) # 32-N_filters applied with the passed kernal_size
    model.add(Activation('relu'))

model.add(Flatten()) # flatten it into 1D after Conv only

# FCL layer
model.add(Dense(128))
model.add(Activation('relu'))

# Output layer - for 10 digits
model.add(Dense(10))
model.add(Activation('softmax'))

model.summary()
```

```python
# We can clearly see with our formula that after convol there is an image reduction to n-m+1 pixels
# We can clearly see that image was 28 x 28 but when it reached conv2d_10 it was reduced to : 22 x 22 pixels
```

```python
# NoW THIS time setup will be same but we will add padding to our convol layers to avoid info loss
```

```python
model = Sequential()

# Add Input 
model.add(Input(shape=(28,28,1)))

n_layers = 3

# Define the 3 convol layers
for _ in range(n_layers):
    model.add(Conv2D(32,kernel_size=(3,3),padding='same')) # 32-N_filters applied with the passed kernal_size
    model.add(Activation('relu'))

model.add(Flatten()) # flatten it into 1D after Conv only

# FCL layer
model.add(Dense(128))
model.add(Activation('relu'))

# Output layer - for 10 digits
model.add(Dense(10))
model.add(Activation('softmax'))

model.summary()
```

```python
# We can clearly see that how after applying padding in each layer ther was no feature loss and our image is still have same amount of information
# image was 28 x 28 and reached conv2d_last it was not reduced and remained : 28 x 28 pixels
```
