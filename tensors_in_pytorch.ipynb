```python
device = torch.device('cuda') # This step enables us to run the tensors using our vgpu
```

A tensor is a multi-dimensional array of data, a geometric object describing linear relations, and a generalization of scalars, vectors, and matrices into higher **dimensions**.

```python
x = torch.tensor([[1,2,3,4,5]],device=device) # A simple tensor in torch where we directly pass values to create a tensor
# This looks like an overhead but is incredibly useful when doing complex tasks on n-dimensional hyperplanes.
```

```python
x
```

```python
type(x) # The type is a tensor
```

```python
x.shape # The shape is 1,5 and is 2d tensor
```

```python
# Do not confuse this with a 1d tensor which are just points in a line :
torch.tensor([1,2,3,4,5]).shape # Notice that this is a vector and is 1D.
```

```python
# And this below is a scaler : 0D
torch.tensor(5).shape # See there is no size ? That is because it is a 0D tensor.
```

```python
# Another examples of higher dimensional tensors are :

# Configuration: [Channels (RGB), Height, Width]
channels = 3
height = 256
width = 256

# Create the 3D tensor with random pixel values between 0 and 1
image_tensor = torch.rand(channels, height, width, device=device)
# We will be covering rand class later for now think of it a tool that creates random numbers by passed dimensions

# Check the dimensions
print(f"Tensor Shape: {image_tensor.shape}")
print(f"Number of dimensions: {image_tensor.ndim}")
```

```python
# An example of 4D tensor is this image being unsqueezed and passed to as batches for convolution.
image_tensor.unsqueeze(0).shape
```

```python
# Now following we will look at some other ways to instantiate matrix , other than implicitly specifying values.
```

```python
# For all these ways we will use a simple size of :
size = (3,3) # Each of this ways share the size parameter
```

```python
torch.empty(size) # Instantiation of an empty tensor with the passed shape
# This function allocates the the allocated size tensor whose values can be anything that already exist on that memory space.
```

```python
torch.zeros(size) # Instantiate a tensor with values filled : 0
```

```python
torch.ones(size) # Instantiate a tensor with values filled : 1
```

```python
torch.rand(size) # Instantiate a tensor with values filled : Random floating numbers
```

```python
torch.tensor((1,2,3)) # Passing a tuple to create a tensor 1D vector
```

```python
torch.eye(5) # Create an idenitity matrix with the passed n value
```

```python
torch.linspace(start=0,end=10,steps=100) # Creates a 2D vector where we pass the starting and the ending point with the step value between these
```

```python
torch.full((2,2),0.5) # Pass in the shape as tuple and we make it such that all the values are instantiated with a single value
```

```python
# We can also do this :
n=3.14
(n)*torch.ones(size,dtype=torch.float32)
```

```python
# Now we can also create using all these but with the dimensions of other tensor
# 1. An existing tensor (e.g., a 2x3 matrix)
existing_tensor = torch.rand(2, 3)

# 2. Create a new tensor with the same shape, filled with 5.5
new_tensor = torch.full_like(existing_tensor, 5.5)

 # We passed in another tensor shape and using that shape we created similar shaped tensor but with our values
existing_tensor.shape,new_tensor.shape
```
